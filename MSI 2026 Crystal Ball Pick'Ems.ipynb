Dựa trên thể thức của MSI 2026, khả năng cao sẽ có 2 đội LCK và 2 đội LPL trong Bracket Stage -> meta LCK/LPL sẽ có ảnh hưởng lớn hơn. Vì vậy, người viết sẽ sử dụng dữ liệu của các giải đấu sau: 
- LCK 2026 Split 1-2
- LCK 2026 Road to MSI
- LPL 2026 Split 2
- LPL 2026 Split 2 Playoffs

Dữ liệu được tải về từ Oracle's Elixir.

# Tướng

```python
import pandas as pd
import numpy as np

lck_playoffs_champ = pd.read_csv("MSI 2026 data\\LCK 2026 Road to MSI - Champion Stats - OraclesElixir.csv")
lck_champ = pd.read_csv("MSI 2026 data\\LCK 2026 Rounds 1-2 - Champion Stats - OraclesElixir.csv")
lpl_playoffs_champ = pd.read_csv("MSI 2026 data\\LPL 2026 Split 2 Playoffs - Champion Stats - OraclesElixir.csv")
lpl_champ = pd.read_csv("MSI 2026 data\\LPL 2026 Split 2 - Champion Stats - OraclesElixir.csv")
```

```python
# Tiền xử lý dữ liệu để tính toán các chỉ số cần thiết
lck_playoffs_number_of_games = 20
lck_number_of_games = 204
lpl_playoffs_number_of_games = 68
lpl_number_of_games = 171

def convert_percentage_to_float(value):
    if isinstance(value, object) and str(value).endswith('%'):
        return float(str(value).strip('%')) / 100
    return value

lck_playoffs_champ['B%'] = lck_playoffs_champ['B%'].apply(convert_percentage_to_float)
lck_champ['B%'] = lck_champ['B%'].apply(convert_percentage_to_float)
lpl_playoffs_champ['B%'] = lpl_playoffs_champ['B%'].apply(convert_percentage_to_float)
lpl_champ['B%'] = lpl_champ['B%'].apply(convert_percentage_to_float)

lck_playoffs_champ.rename(columns={'B%': 'Ban Rate'}, inplace=True)
lck_champ.rename(columns={'B%': 'Ban Rate'}, inplace=True)
lpl_playoffs_champ.rename(columns={'B%': 'Ban Rate'}, inplace=True)
lpl_champ.rename(columns={'B%': 'Ban Rate'}, inplace=True)

lck_playoffs_champ['Games Banned'] = lck_playoffs_champ['Ban Rate'] * lck_playoffs_number_of_games
lck_playoffs_champ['Games Banned'] = lck_playoffs_champ['Games Banned'].apply(lambda x: round(x) if x != '' else 0)
lck_champ['Games Banned'] = lck_champ['Ban Rate'] * lck_number_of_games
lck_champ['Games Banned'] = lck_champ['Games Banned'].apply(lambda x: round(x) if x != '' else 0)
lpl_playoffs_champ['Games Banned'] = lpl_playoffs_champ['Ban Rate'] * lpl_playoffs_number_of_games
lpl_playoffs_champ['Games Banned'] = lpl_playoffs_champ['Games Banned'].apply(lambda x: round(x) if x != '' else 0)
lpl_champ['Games Banned'] = lpl_champ['Ban Rate'] * lpl_number_of_games
lpl_champ['Games Banned'] = lpl_champ['Games Banned'].apply(lambda x: round(x) if x != '' else 0)
```

```python
champ = pd.concat([lck_playoffs_champ, lck_champ, lpl_playoffs_champ, lpl_champ], ignore_index=True)
champ
```

### Làm sạch và tiền xử lý dữ liệu

```python
champ[['W%', 'P%']] = champ[['W%', 'P%']].applymap(convert_percentage_to_float)
champ.rename(columns={'W%': 'Win Rate', 'P%': 'Pick Rate'}, inplace=True)
```

```python
champ['Game Win'] = champ['GP'] * champ['Win Rate']
champ['Game Win'] = champ['Game Win'].apply(lambda x: round(x) if x != '' else 0)
champ
```

```python
champ['K'] = champ['K'].apply(lambda x: int(x) if x != '-' else 0)
```

### Vị tướng nào sẽ được chọn nhiều nhất?

```python
champ[['Champion','GP']].groupby('Champion')['GP'].sum().sort_values(ascending=False).head(15)
```

### Vị tướng nào sẽ có tỉ lệ thắng cao nhất? (được sử dụng trong tối thiểu 5 trận)

```python
champ_above_5_games = champ.groupby('Champion')['GP'].sum().sort_values()
champ_above_5_games_list = champ_above_5_games[champ_above_5_games >= 5].index.tolist()
```

```python
total_games_per_champ = champ[champ['Champion'].isin(champ_above_5_games_list)].where(champ['GP'] > 0).groupby('Champion').agg({'GP': 'sum', 'Game Win': 'sum'})
total_games_per_champ['Win Rate'] = total_games_per_champ['Game Win'] / total_games_per_champ['GP']
total_games_per_champ.sort_values(by='Win Rate', ascending=False)
```

Người viết bỏ qua các tướng như Lissandra, Ornn... do số lượng trận đấu quá ít -> chọn Vayne cho mục này.

### Vị tướng nào sẽ có tỉ lệ thua cao nhất? (được sử dụng trong tối thiểu 5 trận))

```python
total_games_per_champ.sort_values(by='Win Rate', ascending=True)
```

### Vị tướng nào sẽ đạt được nhiều mạng hạ gục nhất?

```python
champ[['Champion','K']].groupby('Champion')['K'].sum().sort_values(ascending=False).head(15)
```

### Vị tướng nào sẽ bị cấm nhiều nhất?

```python
champ[['Champion','Games Banned']].groupby('Champion')['Games Banned'].sum().sort_values(ascending=False).head(15)
```

# Tuyển thủ

```python
lck_playoffs_player = pd.read_csv("MSI 2026 data\\LCK 2026 Road to MSI - Player Stats - OraclesElixir.csv")
lck_player = pd.read_csv("MSI 2026 data\\LCK 2026 Rounds 1-2 - Player Stats - OraclesElixir.csv")
lpl_playoffs_player = pd.read_csv("MSI 2026 data\\LPL 2026 Split 2 Playoffs - Player Stats - OraclesElixir.csv")
lpl_player = pd.read_csv("MSI 2026 data\\LPL 2026 Split 2 - Player Stats - OraclesElixir.csv")

msi_team = ['T1', 'Hanwha Life Esports', 'Top Esports', 'Bilibili Gaming']

player = pd.concat([lck_playoffs_player, lpl_playoffs_player, lck_player, lpl_player], ignore_index=True)
```

```python
player
```

```python
msi_player = player.where(player['Team'].isin(msi_team))
msi_player.dropna(inplace = True, how = 'all')
msi_player
```

### Làm sạch và tiền xử lý dữ liệu

```python
msi_player['FB%'] = msi_player['FB%'].apply(convert_percentage_to_float)
msi_player.rename(columns={'FB%': 'First Blood Rate'}, inplace=True)
```

```python
msi_player['First Blood'] = msi_player['First Blood Rate'] * msi_player['GP']
msi_player['First Blood'] = msi_player['First Blood'].apply(lambda x: round(x) if not pd.isnull(x) else 0)
msi_player
```

### Tại Vòng Khởi Động MSI: Tuyển thủ chuyên nghiệp nào sẽ có KDA cao nhất?

```python
# kiểu éo gì T1 cũng thắng Vòng Khởi Động =))))))
msi_player[msi_player['Team'] == 'T1'].sort_values(by='KDA', ascending=False)
```

### Tuyển thủ nào sẽ có chỉ số lính cao nhất trong một trận đấu?

```python
msi_player.sort_values(by='CSPM', ascending=False)[['Player', 'Team', 'CSPM']]
```

### Tuyển thủ chuyên nghiệp nào sẽ đạt được nhiều Chiến Công Đầu nhất?

```python
msi_player.groupby('Player')['First Blood'].sum().sort_values(ascending=False)
# Dữ liệu về Chiến Công Đầu của các tuyển thủ LPL bị thiếu, nên không thể tính được tổng số Chiến Công Đầu của các tuyển thủ LPL.
```
