- LCK 2026 Split 3-4
- LCK 2026 Season Play-In
- LCK 2026 Season Playoffs
- LPL 2026 Split 3
- LPL 2026 Split Grand Finals

Ngoài ra còn sử dụng dữ liệu của các giải đấu quốc tế tiền CKTG:
- Demacia Cup Global Invitational 2026 (aka cúp C2)
- Worlds Star Challengers Invitational 2026 (aka CKTG tier 2)

Dữ liệu được tải về từ Oracle's Elixir.

# Tướng

```python
import pandas as pd
import numpy as np

lck_playoffs_champ = pd.read_csv("Worlds 2026 data\\LCK 2026 Season Playoffs - Champion Stats - OraclesElixir.csv")
lck_playin_champ = pd.read_csv("Worlds 2026 data\\LCK 2026 Season Play-In - Champion Stats - OraclesElixir.csv")
lck_champ = pd.read_csv("Worlds 2026 data\\LCK 2026 Rounds 3-4 - Champion Stats - OraclesElixir.csv")
lpl_playoffs_champ = pd.read_csv("Worlds 2026 data\\LPL 2026 Grand Finals - Champion Stats - OraclesElixir.csv")
lpl_champ = pd.read_csv("Worlds 2026 data\\LPL 2026 Split 3 - Champion Stats - OraclesElixir.csv")
dcgi_champ = pd.read_csv("Worlds 2026 data\\Demacia Cup Global Invitational 2026 - Champion Stats - OraclesElixir.csv")
wsci_champ = pd.read_csv("Worlds 2026 data\\World Star Challengers Invitational 2026 - Champion Stats - OraclesElixir.csv")
```

```python
# Tiền xử lý dữ liệu để tính toán các chỉ số cần thiết
lck_playoffs_number_of_games = 41
lck_playin_number_of_games = 14
lck_number_of_games = 99
lpl_playoffs_number_of_games = 55
lpl_number_of_games = 167
dcgi_number_of_games = 67 # six seven
wsci_number_of_games = 76 # seven six

def convert_percentage_to_float(value):
    if isinstance(value, object) and str(value).endswith('%'):
        return float(str(value).strip('%')) / 100
    return value

lck_playoffs_champ['B%'] = lck_playoffs_champ['B%'].apply(convert_percentage_to_float)
lck_playin_champ['B%'] = lck_playin_champ['B%'].apply(convert_percentage_to_float)
lck_champ['B%'] = lck_champ['B%'].apply(convert_percentage_to_float)
lpl_playoffs_champ['B%'] = lpl_playoffs_champ['B%'].apply(convert_percentage_to_float)
lpl_champ['B%'] = lpl_champ['B%'].apply(convert_percentage_to_float)
dcgi_champ['B%'] = dcgi_champ['B%'].apply(convert_percentage_to_float)
wsci_champ['B%'] = wsci_champ['B%'].apply(convert_percentage_to_float)

lck_playoffs_champ.rename(columns={'B%': 'Ban Rate'}, inplace=True)
lck_playin_champ.rename(columns={'B%': 'Ban Rate'}, inplace=True)
lck_champ.rename(columns={'B%': 'Ban Rate'}, inplace=True)
lpl_playoffs_champ.rename(columns={'B%': 'Ban Rate'}, inplace=True)
lpl_champ.rename(columns={'B%': 'Ban Rate'}, inplace=True)
dcgi_champ.rename(columns={'B%': 'Ban Rate'}, inplace=True)
wsci_champ.rename(columns={'B%': 'Ban Rate'}, inplace=True)

lck_playoffs_champ['Games Banned'] = lck_playoffs_champ['Ban Rate'] * lck_playoffs_number_of_games
lck_playoffs_champ['Games Banned'] = lck_playoffs_champ['Games Banned'].apply(lambda x: round(x) if x != '' else 0)
lck_playin_champ['Games Banned'] = lck_playin_champ['Ban Rate'] * lck_playin_number_of_games
lck_playin_champ['Games Banned'] = lck_playin_champ['Games Banned'].apply(lambda x: round(x) if x != '' else 0)
lck_champ['Games Banned'] = lck_champ['Ban Rate'] * lck_number_of_games
lck_champ['Games Banned'] = lck_champ['Games Banned'].apply(lambda x: round(x) if x != '' else 0)
lpl_playoffs_champ['Games Banned'] = lpl_playoffs_champ['Ban Rate'] * lpl_playoffs_number_of_games
lpl_playoffs_champ['Games Banned'] = lpl_playoffs_champ['Games Banned'].apply(lambda x: round(x) if x != '' else 0)
lpl_champ['Games Banned'] = lpl_champ['Ban Rate'] * lpl_number_of_games
lpl_champ['Games Banned'] = lpl_champ['Games Banned'].apply(lambda x: round(x) if x != '' else 0)
dcgi_champ['Games Banned'] = dcgi_champ['Ban Rate'] * dcgi_number_of_games
dcgi_champ['Games Banned'] = dcgi_champ['Games Banned'].apply(lambda x: round(x) if x != '' else 0)
wsci_champ['Games Banned'] = wsci_champ['Ban Rate'] * wsci_number_of_games
wsci_champ['Games Banned'] = wsci_champ['Games Banned'].apply(lambda x: round(x) if x != '' else 0)
```

```python
champ = pd.concat([lck_playoffs_champ, lck_playin_champ, lck_champ, lpl_playoffs_champ, lpl_champ, dcgi_champ, wsci_champ], ignore_index=True)
champ
```

### Làm sạch và tiền xử lý dữ liệu

```python
champ[['W%', 'P%']] = champ[['W%', 'P%']].applymap(convert_percentage_to_float)
champ.rename(columns={'W%': 'Win Rate', 'P%': 'Pick Rate'}, inplace=True)
```

```python
champ['Game Win'] = champ['GP'] * champ['Win Rate']
champ['Game Win'] = champ['Game Win'].apply(lambda x: round(x) if x != '' else 0)
champ
```

```python
# pick/ban presence = games played + games banned
champ['Presence'] = champ['GP'] + champ['Games Banned']
champ
```

```python
champ['K'] = champ['K'].apply(lambda x: int(x) if x != '-' else 0)
champ['D'] = champ['D'].apply(lambda x: int(x) if x != '-' else 0)
```

### Vị tướng nào sẽ được cấm/chọn nhiều nhất?

```python
champ[['Champion','Presence']].groupby('Champion')['Presence'].sum().sort_values(ascending=False).head(15)
```

### Vị tướng nào sẽ có tỉ lệ thắng cao nhất? (được sử dụng trong tối thiểu 5 trận)

```python
champ_above_5_games = champ.groupby('Champion')['GP'].sum().sort_values()
champ_above_5_games_list = champ_above_5_games[champ_above_5_games >= 5].index.tolist()
```

```python
total_games_per_champ = champ[champ['Champion'].isin(champ_above_5_games_list)].where(champ['GP'] > 0).groupby('Champion').agg({'GP': 'sum', 'Game Win': 'sum'})
total_games_per_champ['Win Rate'] = total_games_per_champ['Game Win'] / total_games_per_champ['GP']
total_games_per_champ.sort_values(by='Win Rate', ascending=False)
```

Người viết bỏ qua Nidalee do số lượng trận đấu quá ít -> chọn Cassiopeia cho mục này.

### Vị tướng nào sẽ có số mạng nằm xuống trung bình mỗi ván đấu cao nhất? (được sử dụng trong tối thiểu 5 trận))

```python
avg_death_per_champ = champ[champ['Champion'].isin(champ_above_5_games_list)].where(champ['GP'] > 0).groupby('Champion').agg({'GP': 'sum', 'D': 'sum'})
avg_death_per_champ['Avg Death'] = avg_death_per_champ['D'] / avg_death_per_champ['GP']
```

```python
avg_death_per_champ.sort_values(by='Avg Death', ascending=False)
```

Người viết bỏ qua Pyke do số lượng trận đấu quá ít -> chọn Nautilus cho mục này.

# Tuyển thủ

```python
lck_playoffs_player = pd.read_csv("Worlds 2026 data\\LCK 2026 Season Playoffs - Player Stats - OraclesElixir.csv")
lck_playin_player = pd.read_csv("Worlds 2026 data\\LCK 2026 Season Play-In - Player Stats - OraclesElixir.csv")
lck_player = pd.read_csv("Worlds 2026 data\\LCK 2026 Rounds 3-4 - Player Stats - OraclesElixir.csv")
lpl_playoffs_player = pd.read_csv("Worlds 2026 data\\LPL 2026 Grand Finals - Player Stats - OraclesElixir.csv")
lpl_player = pd.read_csv("Worlds 2026 data\\LPL 2026 Split 3 - Player Stats - OraclesElixir.csv")

worlds_team = ['T1', 'Hanwha Life Esports', 'Dplus Kia', 'Gen.G', 'Top Esports', 'Bilibili Gaming', 'Invictus Gaming', "Anyone's Legend"]

player = pd.concat([lck_playoffs_player, lpl_playoffs_player, lck_playin_player, lck_player, lpl_player], ignore_index=True)
```

```python
player
```

```python
worlds_player = player.where(player['Team'].isin(worlds_team))
worlds_player.dropna(inplace = True, how = 'all')
worlds_player
```

### Làm sạch và tiền xử lý dữ liệu

```python
worlds_player['FB%'] = worlds_player['FB%'].apply(convert_percentage_to_float)
worlds_player.rename(columns={'FB%': 'First Blood Rate'}, inplace=True)
```

```python
worlds_player['First Blood'] = worlds_player['First Blood Rate'] * worlds_player['GP']
worlds_player['First Blood'] = worlds_player['First Blood'].apply(lambda x: round(x) if not pd.isnull(x) else 0)
worlds_player
```

### Tuyển thủ chuyên nghiệp nào sẽ có KDA cao nhất?

```python
worlds_player_kda = worlds_player.groupby('Player').agg({'K': 'sum', 'D': 'sum', 'A': 'sum'})
worlds_player_kda['agg_KDA'] = (worlds_player_kda['K'] + worlds_player_kda['A']) / (worlds_player_kda['D'] + 1)  # Adding 1 to avoid division by zero
worlds_player_kda.sort_values(by='agg_KDA', ascending=False)
```

### Tuyển thủ nào sẽ có trung bình sát thương trong một trận đấu cao nhất?

```python
worlds_player.sort_values(by='DPM', ascending=False)[['Player', 'Team', 'DPM']]
# JiaQi is in a weaker team, so he might be farming Nirvana group opponents.
```
